**Research question**

How can we use available search performance data to prioritize content pages for review, while considering potential opportunities and the costs of incorrect recommendations?

**Unit of analysis:** One content page (or page-level observation, depending on the dataset structure).

**Decision:** Which pages should a content team review first when time and resources are limited?

**Output:** An explainable priority score and a ranked list of pages that may benefit from further investigation.

**Action:** A content strategist or SEO specialist reviews high-priority pages, checks the context, and decides whether a content update is justified.

**Cost of a wrong recommendation:**
- False positive: The team spends time reviewing or updating a page that did not need attention.
- False negative: A page with a potentially valuable improvement opportunity is overlooked.
- Additional risk: Unnecessary changes could negatively affect existing content performance.

**Why data or ML can help:** Search performance involves several interacting signals, including clicks, impressions, and trends. Data analysis can help identify patterns and compare pages consistently. ML may support prioritization at scale, but its usefulness must be tested against simpler rule-based approaches.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python


import pandas as pd

url = "https://raw.githubusercontent.com/aminaljevakovich/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Dataset shape:", df.shape)

print("\nAvailable columns:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())
```

```python

# Explore available performance-related columns
performance_cols = [
    col for col in df.columns
    if any(term in col.lower() for term in
           ["click", "impression", "trend", "declin", "position"])
]

print("Performance-related columns:")
print(performance_cols)

print("\nNumber of unique content IDs:")
print(df["content_id"].nunique())

print("\nNumber of unique clients:")
print(df["client_id"].nunique())

print("\nSummary of performance metrics:")
display(df[performance_cols].describe(include="all").T)
```

```python

# Compare clicks across two 30-day periods
declining = df["clicks_last_30d"] < df["clicks_prev_30d"]

declining_count = declining.sum()
declining_pct = declining.mean() * 100

print(f"Pages with declining clicks: {declining_count:,}")
print(f"Percentage of pages with declining clicks: {declining_pct:.2f}%")
print(f"Average clicks, previous 30 days: {df['clicks_prev_30d'].mean():.2f}")
print(f"Average clicks, last 30 days: {df['clicks_last_30d'].mean():.2f}")
```

**Initial data exploration — Refresh / Content Opportunity Scoring**

## 4. Careful words: what I can and can't claim

**What I can claim**

My analysis can describe observed patterns in anonymized search performance data and identify pages that may deserve further investigation. It can support directional, evidence-based recommendations and help content teams prioritize limited review resources.

The proposed ranking should be interpreted as a decision-support tool rather than an automatic instruction to update content. Its usefulness must be evaluated against simpler approaches.

**What I cannot claim**

I cannot conclude that declining clicks are necessarily caused by outdated or low-quality content. I also cannot claim that updating a recommended page will increase traffic, improve Google rankings, or produce a specific business outcome.

The available data are observational and may be affected by seasonality, search demand, competition, and other unmeasured factors. Any ML results will require validation and careful interpretation before they can support operational decisions.

```python

# Check the evidence behind the research framing

total_pages = len(df)
declining_pages = (
    df["clicks_last_30d"] < df["clicks_prev_30d"]
).sum()

print("Evidence check")
print(f"Total pages analyzed: {total_pages:,}")
print(f"Pages with fewer clicks: {declining_pages:,}")
print(f"Share with fewer clicks: {declining_pages / total_pages:.2%}")

print("\nInterpretation:")
print("A decline in clicks is an observed pattern.")
print("It does not prove that a content refresh will improve performance.")
```
